```python
import random
print("welcome to the Number Guessing Game!")
while True:
    secret_number = random.randint(1, 100)
    max_attempts = 7
    print("\nI am thinking of a number between 1 and 100")
    print("You have", max_attempts, "attempts")
    for attempt in range(1, max_attempts + 1):
        try:
            print("attempt", attempt)
            guess = int(input("Enter your guess: "))
            if guess == secret_number:
                print("Correct. You guessed the number in", attempt, "attempts.")
                break
            elif guess > secret_number:
                print("Too High!")
            else:
                print("Too Low!")
            print("remaining Attempts:", max_attempts - attempt)
        except ValueError:
            print("invalid input! Please enter a number.")
    else:
        print("\ngame Over!")
        print("the correct number was:", secret_number)
    choice = input("\nDo you want to play again? (yes/no): ")
    if choice.lower() != "yes":
        print("Thanks for playing!")
        break
```
